# Morning Drill

A live sandbox. Run it top to bottom each morning and overwrite it — nothing
here is a record. The record is `drill_records.json`, written as you go, and
the progress agent reads it before setting anything.

Anything below from yesterday is scratch.

## While you work

    session.ask("...")     a question, without submitting an attempt
    session.reference()    the worked solution — refuses until you have passed
    session.status()       progress, and what you have done today

## Finish

Write down anything the score cannot see — what felt shaky, what you had to
look up, what you want more of. The progress agent reads these notes and
weighs them against the pass rate, so a quiet "merges still feel unfamiliar"
changes what you get tomorrow.

Then run it. This writes the session to `drill_records.json`.

```python
session.save(notes="")
```
